1. Quais as diferentes arquiteturas de microsserviços?

A arquitetura de microserviços pode ser implementada de várias formas, dependendo do domínio, da forma de comunicação e da maneira como os serviços colaboram, sendo modelos consolidados de organização utilizados para definir como os serviços se comportam, interagem e são estruturados. A seguir, estão os principais estilos de arquitetura adotados no desenvolvimento de microserviços,

**1. Arquitetura Orientada a Domínio (DDD / Bounded Contexts),**
Nesse estilo, os microserviços são definidos com base nos *subdomínios do negócio*, seguindo o conceito de *bounded contexts* do Domain-Driven Design, cada serviço sendo responsável por uma parte específica e isolada do domínio, mantendo alta coesão e baixo acoplamento. É o modelo mais utilizado para decompor sistemas complexos,

**2. Arquitetura REST-Based (Comunicação Síncrona via HTTP),**
Nessa abordagem, cada microserviço expõe uma *API REST*, comunicando-se de forma síncrona via HTTP/JSON, sendo o estilo mais tradicional e simples de implementar, adequado para operações CRUD e interações diretas entre serviços,

**3. Arquitetura Event-Driven (Assíncrona por Eventos),**
Os microserviços se comunicam por meio de eventos utilizando tecnologias como Kafka, RabbitMQ ou SQS, com três variações comuns:
• *Event Notification*, quando o evento apenas notifica que algo ocorreu,
• *Event-Carried State Transfer*, quando o evento carrega dados usados por outros serviços,
• *Event Sourcing*, quando todo o estado é reconstruído a partir de eventos persistidos,
Esse modelo proporciona maior desacoplamento, reduz bloqueios e melhora desempenho e escalabilidade,

**4. Arquitetura Coreografada e Arquitetura Orquestrada,**
Esse estilo define como fluxos distribuídos entre microserviços são coordenados, podendo ocorrer por:
• *Coreografia*, em que não há um serviço central e cada microserviço reage a eventos,
• *Orquestração*, em que um componente orquestrador controla o fluxo, como Saga Orchestrator, Camunda ou Temporal,
Esse tipo de arquitetura é comum em sistemas com múltiplas etapas de negócio distribuídas entre vários serviços,

**5. Arquitetura Service Mesh (Sidecar Pattern),**
A comunicação entre microserviços é gerenciada por proxies sidecar, como Envoy, utilizados com plataformas como Istio ou Linkerd, fornecendo recursos como mTLS automático, retries, circuit breaking, observabilidade e controle centralizado de tráfego, sendo especialmente útil quando o sistema cresce e a complexidade da comunicação aumenta,

**6. Arquitetura Serverless + Microsserviços,**
Os serviços são implementados como funções serverless, como AWS Lambda ou Azure Functions, permitindo alta escalabilidade, funcionamento efêmero e integração natural com arquiteturas orientadas a eventos, sendo uma alternativa moderna para workloads que exigem elasticidade extrema.

**API Gateway** é um padrão arquitetural utilizado em microserviços que funciona como uma camada intermediária entre o cliente e os diversos serviços da aplicação. Ele atua como um ponto de entrada único, recebendo as requisições do cliente e encaminhando-as para os microserviços responsáveis.

Esse padrão centraliza e gerencia quais serviços devem ser chamados, em qual ordem e sob quais condições. Dessa forma, o cliente não precisa conhecer a estrutura interna da aplicação nem fazer múltiplas chamadas para serviços diferentes.

A principal vantagem do API Gateway é a praticidade: ele unifica o acesso e simplifica o consumo da aplicação. Com isso, evita que o cliente tenha de se comunicar diretamente com cada microserviço individualmente, reduzindo a complexidade e melhorando a experiência de integração.

**Backends for Frontends (BFF)** é um padrão arquitetural utilizado em microserviços que se assemelha ao API Gateway, mas com uma abordagem mais especializada. Em vez de um único ponto de entrada para todos os tipos de clientes, o BFF propõe a criação de backends distintos para cada frontend da aplicação.

Nessa lógica, cada interface, como web, mobile ou desktop, possui um backend próprio, projetado especificamente para atender às necessidades e características daquele tipo de cliente. Isso permite que cada frontend receba exatamente os dados e funcionalidades de que precisa, sem acúmulo de lógica adicional ou ajustes complexos.

A principal vantagem do BFF é a customização: cada frontend interage com um backend otimizado para sua experiência, resultando em uma aplicação mais eficiente, objetiva e alinhada às expectativas do usuário.

Abaixo diagrama com a comparação de ambos modelos arquiteturais:




**Service Registry (ou Service Discovery)** é um serviço interno da aplicação responsável por manter uma lista atualizada dos microserviços ativos. Ele pode ser consultado tanto pelo API Gateway quanto pelos próprios microserviços quando precisam acessar outro serviço. Assim, quando um serviço X precisa se comunicar com o serviço Y, ele consulta o Service Registry para descobrir onde o serviço Y está rodando naquele momento e então realiza a requisição diretamente. Esse padrão é comum em arquiteturas que exigem escalabilidade dinâmica e lidam com um grande volume de serviços distribuídos. Seu papel é fornecer dados precisos e atualizados sobre a infraestrutura, permitindo descoberta automática e garantindo que os serviços se encontrem mesmo em ambientes com escalabilidade dinâmica e diversas instâncias distribuídas.

**Circuit Breaker** é um padrão arquitetural utilizado em microserviços para evitar que falhas contínuas em um serviço comprometem toda a aplicação. Ele monitora as requisições feitas entre os microserviços e identifica quando um serviço começa a falhar repetidamente. Ao detectar um número elevado de erros, o Circuit Breaker “abre o circuito”, interrompendo chamadas futuras para aquele serviço até que ele volte a operar normalmente.

Na prática, funciona da seguinte forma: imagine um serviço A que depende dos serviços B, C e D. Se o serviço D começa a falhar de forma consistente, o Circuit Breaker entra em ação e impede que o serviço A continue fazendo requisições ao serviço D. Isso evita sobrecarga, timeouts excessivos e protege a aplicação de efeitos em cascata.

Durante o período em que o circuito está aberto, o Circuit Breaker pode realizar testes periódicos, enviando chamadas de verificação, para identificar se o serviço D se recuperou. Assim que ele responder corretamente, o circuito é “fechado” novamente, permitindo que o fluxo normal de requisições seja restabelecido.

Esse padrão é essencial para garantir a resiliência da arquitetura, impedindo que falhas localizadas provoquem indisponibilidade generalizada ou degradação do desempenho em larga escala.



**Bulkhead** funciona como um isolador de recursos dentro de um sistema. Ele estabelece um limite de requisições e garante que, se dentro de um mesmo serviço ocorrer um overload de requisições para um recurso específico, apenas aquele recurso ficará indisponível ou sobrecarregado, sem afetar os demais recursos do mesmo serviço, nem outros serviços do sistema.

Por exemplo, imagine um serviço X que possui os recursos A, B, C e D; Se o recurso B receber mais requisições do que sua capacidade suporta, ele ficará saturado e deixará de atender novas chamadas. O Bulkhead garante que essa saturação de B não se espalhe, os recursos A, C e D continuarão funcionando normalmente. Assim, apenas as requisições que dependem do recurso B são afetadas, enquanto o restante do serviço e do sistema continuam operando.

**Retry Pattern** é um padrão arquitetural utilizado para lidar com falhas temporárias em sistemas distribuídos. Quando uma requisição falha, o Retry Pattern realiza novas tentativas automaticamente, seguindo regras pré-definidas.

Essas regras podem incluir o número máximo de tentativas, intervalos entre cada retry e estratégias de espera, como tempos crescentes (por exemplo, aguardar 1 minuto, depois 2, depois 3). Caso todas as tentativas falhem, o padrão retorna um erro final ou aciona outros mecanismos de resiliência, como o Circuit Breaker.

Esse padrão é amplamente utilizado para garantir que o sistema continue operando mesmo diante de falhas momentâneas, aumentando a robustez e a confiabilidade da arquitetura.

**Saga Pattern** é um padrão arquitetural utilizado para coordenar transações distribuídas em sistemas baseados em microserviços, garantindo consistência de dados mesmo quando uma operação envolve múltiplos serviços independentes.

Em arquiteturas monolíticas, uma transação que envolve várias etapas normalmente utiliza mecanismos tradicionais de *commit* e *rollback*: ou todas as operações são confirmadas, ou todas são desfeitas. Já em ambientes distribuídos, esse tipo de controle centralizado não é possível, pois não existe uma única transação atômica envolvendo diversos microserviços. É nesse contexto que o Saga Pattern atua.

Ele define uma sequência de passos, onde cada serviço realiza sua operação e, em caso de falha em algum ponto, são disparadas operações compensatórias para desfazer as alterações já realizadas garantindo consistência no fluxo completo.

Existem duas formas principais de implementação:

**1. Orchestration (orquestração)**
Nesse modelo, existe um serviço central, chamado de *orchestrator*, que controla toda a lógica da saga. Ele coordena cada etapa, chama explicitamente cada microserviço na ordem correta e decide o que fazer em caso de falha, disparando as ações compensatórias quando necessário.

**2. Choreography (coreografia)**
Na coreografia, não existe um controlador central. Cada microserviço executa sua tarefa e, ao terminar, publica um evento informando o resultado. Os demais microserviços “ouvem” esses eventos e decidem como reagir. Se ocorrer uma falha, cada serviço é responsável por publicar eventos de compensação, desfazendo as etapas anteriores.

Um exemplo comum está no domínio bancário: ao realizar uma transferência, diversos serviços entram na operação, validação de dados, verificação de saldo, débito na conta de origem, crédito na conta destino, geração de comprovante etc. Se algum desses passos falhar, o Saga Pattern garante que as operações anteriores sejam revertidas, restaurando o estado original e evitando inconsistências como valores debitados sem crédito correspondente.

Assim, o Saga Pattern torna possíveis transações seguras, consistentes e coordenadas em arquiteturas distribuídas, onde não existe um único sistema responsável pela transação inteira.

**Event Sourcing** é um padrão arquitetural que registra todas as alterações ocorridas em um sistema na forma de eventos imutáveis, preservando o histórico completo das mudanças ao invés de armazenar apenas o estado final dos dados.

Em sistemas tradicionais, quando uma atualização ocorre, o estado anterior é sobrescrito, mantendo-se apenas a versão mais recente. No Event Sourcing, cada mudança relevante gera um novo evento, e todos esses eventos são registrados sequencialmente em um Event Store ou banco de dados apropriado. Dessa forma, a aplicação não guarda somente o resultado final, mas cada passo que levou até ele.


**CQRS (Command Query Responsibility Segregation)** é um padrão arquitetural que separa a aplicação em dois modelos distintos: um especializado em escrita (commands) e outro especializado em leitura (queries). Em vez de usar um único modelo para realizar ambos os tipos de operação, o CQRS promove a separação explícita dessas responsabilidades.

No modelo de comandos, são processadas ações que modificam o estado da aplicação, como criar, atualizar ou excluir dados. Já o modelo de consulta é otimizado exclusivamente para a leitura, permitindo respostas rápidas e eficientes sem carregar as restrições e validações típicas das operações de escrita.

Na maioria dos sistemas, as operações de leitura são muito mais frequentes do que as de escrita. Por isso, o CQRS melhora a performance geral, simplifica o modelo de leitura e permite escalar cada lado independentemente, resultando em um sistema mais eficiente e robusto.

**Database Per Service** é um padrão arquitetural em que **cada microserviço possui seu próprio banco de dados**, totalmente independente e isolado dos demais serviços do sistema. Isso significa que o serviço **X** tem sua própria base, o serviço **Y** a sua, o serviço **Z** a sua, e assim por diante.

Essa independência garante que cada serviço seja **únicamente responsável por seus próprios dados**, mantendo controle total sobre seu modelo, suas regras e sua forma de persistência. Como consequência, os serviços ficam se mantêm **desacoplados**, facilitando a evolução individual de cada um sem impactar os outros.

**Data Sharding** é um padrão arquitetural que consiste em dividir um conjunto muito grande de dados em partes menores, distribuídas entre diferentes servidores ou bancos de dados. Em vez de concentrar todos os registros em um único banco gigantesco, como um banco de dados contendo clientes do mundo inteiro, o sistema passa a trabalhar com múltiplos bancos, cada um responsável por uma fatia específica desses dados.

Essa divisão pode seguir diversos critérios, sendo comum o uso de intervalos definidos. Por exemplo:

* O **Shard 1** armazena clientes com IDs de 1 até 1.000.000
* O **Shard 2** cuida dos registros de 1.000.001 até 2.000.000
* O **Shard 3** dos IDs seguintes, e assim por diante

Com isso, cada shard processa somente sua parcela do conjunto total, diminuindo carga, aumentando a escalabilidade e melhorando o desempenho geral.

As requisições são direcionadas ao shard correto por meio de uma regra de roteamento, como intervalos numéricos, hashing ou outra estratégia determinística, garantindo que a aplicação saiba exatamente para qual banco enviar a consulta, sem a necessidade de verificar todos os shards.

**Polyglot Persistence** é um padrão arquitetural no qual cada serviço utiliza o tipo de banco de dados que melhor atende às suas próprias necessidades. Em vez de forçar toda a aplicação a usar um único mecanismo de persistência, cada serviço pode escolher tecnologias distintas, como bancos relacionais, NoSQL, grafos, time-series ou mecanismos de busca, de acordo com suas características e demandas.

Esse modelo permite que cada parte do sistema opere com o armazenamento mais eficiente para seu caso de uso, garantindo maior desempenho, flexibilidade e independência entre os serviços. Assim, a aplicação como um todo torna-se mais sustentável, otimizada e preparada para lidar com diferentes cenários de dados.

**Sidecar Pattern** é um padrão arquitetural no qual um serviço auxiliar é executado ao lado do serviço principal, compartilhando o mesmo ambiente de execução e o mesmo ciclo de vida. Esse serviço complementar fornece funcionalidades de suporte, como monitoramento, logging, segurança, comunicação, configuração ou proxy de rede, permitindo que o serviço principal permaneça focado exclusivamente na sua lógica de negócio.

O sidecar não precisa conhecer os detalhes internos do serviço principal, pois sua função não é executar a regra de negócio, mas sim oferecer capacidades adicionais essenciais para o funcionamento saudável do sistema. Ele opera de forma independente, mas coordenada: quando o serviço principal inicia, o sidecar inicia; quando o serviço principal é interrompido, o sidecar também é.

Esse padrão promove modularidade, facilita a padronização de funcionalidades transversais e reduz a complexidade do código do serviço principal, tornando a arquitetura mais organizada, observável e sustentável.

**Smart Endpoints & Dumb Pipes** é um princípio arquitetural que orienta o design de sistemas distribuídos, especialmente microserviços. Ele defende que **a inteligência da aplicação deve estar nas extremidades**, nos serviços e seus endpoints, enquanto **os meios de comunicação entre eles devem permanecer simples e leves**.

Os smart endpoints são responsáveis por concentrar a lógica de negócio, filtragem, validação, transformação de dados e qualquer outra regra necessária para que o serviço cumpra seu papel de forma clara e eficiente. Assim, cada microserviço atua de maneira autônoma, encapsulando suas próprias responsabilidades e permanecendo altamente desacoplado dos demais.

As dumb pipes, por outro lado, representam canais de comunicação minimalistas, como HTTP/REST, mensageria simples ou filas básicas, que não realizam processamento complexo nem aplicam lógica adicional. Eles apenas transportam as mensagens, sem tentar interpretar ou manipular seu conteúdo.

Esse padrão ajuda a manter o sistema mais flexível, modular e fácil de evoluir, garantindo que a complexidade esteja dentro dos microserviços, e não no caminho que conecta um ao outro.

**Asynchronous Messaging & Event-Driven Architecture** é um padrão arquitetural baseado na comunicação assíncrona entre serviços. Em vez de um serviço enviar uma requisição e ficar aguardando uma resposta imediata, ele simplesmente **dispara um evento** e continua executando suas tarefas. A resposta chegará mais tarde, por meio de uma fila ou mecanismo de mensageria.

Nesse modelo, os serviços não ficam bloqueados esperando resultados, o que reduz atrasos, melhora a escalabilidade e evita gargalos. Cada evento enviado é processado de forma independente e, quando concluído, sua resposta é armazenada em uma fila ou tópico para ser consumida pelo serviço interessado.

Esse tipo de arquitetura melhora a eficiência geral do sistema, pois permite que serviços funcionem de maneira desacoplada e responsiva, processando eventos conforme eles chegam, sem a necessidade de sincronização imediata entre produtor e consumidor.

**Consumer Driven Contracts (CDC)** é uma abordagem de integração entre serviços que parte do princípio de que o consumidor da API define o contrato que descreve exatamente o que ele espera do serviço provedor. Em vez de o provedor impor o formato da resposta, é o consumidor quem descreve quais campos, tipos e estruturas são necessários para que ele funcione corretamente.

Nesse modelo, imagine que o **Serviço A** consome a API do **Serviço B**. O Serviço A cria um contrato especificando requisitos como: existir um objeto JSON com determinado campo `x` do tipo *string* e um campo `y` do tipo *number*. Esse contrato é armazenado em arquivos que representam as expectativas formais do consumidor.

O Serviço B, por sua vez, utiliza esse contrato em seus próprios testes, chamados de tests de pacto, para garantir que sua API continue fornecendo exatamente o que os consumidores esperam. Se alguma mudança no Serviço B quebrar esse contrato, os testes falham imediatamente, evitando integrações quebradas em produção.

Esse mecanismo é especialmente valioso em ambientes de microserviços mantidos por equipes diferentes. Ele permite que cada equipe evolua seu serviço com segurança, assegurando que as dependências externas permaneçam funcionando conforme o esperado, e preservando a confiança e a estabilidade entre consumidores e provedores.


2. Quando não devemos utilizar microsserviços?

A arquitetura de microserviços não é indicada para aplicações pequenas ou com requisitos simples, ela foi criada para resolver problemas de escalabilidade, complexidade organizacional e alto volume de usuários, típicos de sistemas de grande porte, como bancos, plataformas globais, e serviços de streaming.

Quando um sistema não exige escalabilidade massiva, não terá um crescimento acelerado, ou não possui uma operação complexa, o custo de adotar microserviços supera os benefícios, pois esse modelo exige infraestrutura mais elaborada, monitoramento distribuído, orquestração e automação, cultura DevOps madura, equipes com maior especialização, e maior complexidade para testes, deploy e comunicação interna.

Por isso, em projetos de pequeno porte, o mais adequado é começar com uma arquitetura monolítica simples, que é mais rápida de desenvolver, mais barata de manter e menos complexa, e somente migrar para microserviços caso a aplicação realmente cresça ou enfrente problemas que justifiquem essa mudança.


3. O que caracteriza uma aplicação nativa da nuvem?

Uma aplicação nativa em nuvem é projetada desde o início para aproveitar ao máximo o ambiente de cloud. Ela se baseia em quatro pilares principais:

1. Responsividade (Responsiveness)
A aplicação responde de forma rápida, consistente e estável, mesmo sob variação de carga. Ela consegue manter boa performance independente do fluxo de usuários.

2. Resiliência (Resilience)
Ela continua funcionando mesmo quando partes do sistema falham. Falhas são esperadas e tratadas com mecanismos como retries, circuit breakers, redundância e isolamento.

3. Elasticidade (Elasticity)
A aplicação escala automaticamente para cima ou para baixo conforme a demanda, otimizando custo e uso de recursos — algo nativo da computação em nuvem.

4. Comunicação orientada a mensagens (Message-driven)
A interação entre componentes é assíncrona e baseada em eventos ou filas, o que reduz acoplamento e aumenta tolerância a falhas.

Relação com microserviços

A arquitetura de microserviços se encaixa perfeitamente nesses princípios porque ela naturalmente favorece resiliência, escalabilidade, isolamento e comunicação assíncrona. Por isso microserviços são considerados um estilo arquitetural cloud native por natureza.

4. Liste as práticas da Twelve-Factor App.

Em serviços distribuídos, a atenção à esses fatores deve ser redobrada.

* Base de código — Uma única base de código rastreada por controle de versão, com múltiplos deploys.
* Dependências — Declare e isole as dependências.
* Configurações — Armazene as configurações no ambiente.
*  Serviços de apoio — Trate serviços de apoio como recursos anexados.
* Build, release, run — Separe estritamente os estágios de build, release e execução.
* Processos — Execute a aplicação como um ou mais processos stateless.
* Vínculo de porta — Exporte serviços via ligação de porta.
* Concorrência — Escalone por meio de processos.
* Descartabilidade — Maximize a robustez com inicialização e desligamento rápidos.
* Ambientes dev/prod semelhantes — Mantenha desenvolvimento, teste e produção o mais semelhantes possível.
* Logs — Trate logs como fluxos de eventos.
* Processos administrativos — Execute tarefas administrativas ou de manutenção como processos pontuais.

5. Quais as principais diferenças entre o estilo arquitetura Monolítica e o estilo arquitetura de Microsserviços?


**Arquitetura Monolítica**

É uma aplicação única, empacotada e implantada como um único deploy. Todos os módulos compartilham o mesmo espaço de execução, o que simplifica o desenvolvimento inicial, a manutenção e o gerenciamento. É mais indicada para projetos de pequeno porte, com complexidade moderada e pouca necessidade de escalabilidade avançada. Sua escalabilidade é tipicamente vertical, aumentando os recursos do servidor inteiro como um bloco único.

**Arquitetura de Microserviços**

Divide o sistema em serviços pequenos, independentes e implantáveis de maneira separada. É usada principalmente em projetos de médio e grande porte, onde há exigência de maior desempenho, escalabilidade, flexibilidade e tolerância a falhas. Os microserviços favorecem resiliência, escalabilidade, isolamento e comunicação assíncrona, e se beneficiam fortemente da computação em nuvem. Sua escalabilidade é horizontal, aumentando ou reduzindo apenas os serviços necessários, conforme a demanda.

6. Quais as principais diferenças entre Kubernetes e Minikube?

O **Minikube** é voltado para desenvolvimento local. Ele roda um único cluster Kubernetes em uma única máquina — geralmente por meio de uma máquina virtual — permitindo ao desenvolvedor testar novas funcionalidades, aprender como o Kubernetes funciona e validar aplicações de forma simples. Além disso, ele oferece recursos que facilitam o uso, como o Kubernetes Dashboard, que ajuda a visualizar objetos e acompanhar o comportamento do cluster.

Já o **Kubernetes** é a plataforma completa de orquestração de contêineres usada em ambientes reais, podendo trabalhar com vários nós distribuídos. Ele é projetado para alta disponibilidade, escalabilidade, resiliência a falhas e para executar diversas aplicações em larga escala. Enquanto o Minikube serve para estudo e testes, o Kubernetes é utilizado para rodar sistemas de produção que exigem robustez e desempenho.

7. Em quais cenários é mais indicado usar Kubernetes ?

O Kubernetes é mais adequado para ambientes de produção, onde existe a necessidade de gerenciar várias aplicações, múltiplos nós e lidar com tráfego real. Ele é projetado para garantir alta disponibilidade, escalabilidade, tolerância a falhas e orquestração eficiente de contêineres em larga escala. Por isso, é a escolha natural quando o objetivo é rodar sistemas completos e robustos, envolvendo diversos serviços e demandas em crescimento.

9. Descreva as principais vantagens de usar Reactive Spring para desenvolver microserviços.

O Reactive Spring é recomendado para aplicações que precisam lidar com fluxos intensos de dados de maneira eficiente. Ele segue um paradigma assíncrono, não bloqueante e baseado em eventos, permitindo que o sistema processe um grande número de requisições simultaneamente sem travamentos.

Um dos seus principais recursos é o backpressure, que controla o ritmo de entrada de dados e impede que o sistema seja sobrecarregado, garantindo estabilidade mesmo durante picos de tráfego. Essa abordagem torna o Reactive Spring ideal para aplicações que recebem milhares de requisições, como APIs de alta demanda, serviços de streaming, mensageria e sistemas que operam com volume contínuo de eventos.

É importante destacar que o Reactive Spring oferece benefícios reais especialmente em aplicações que fazem uso pesado de operações de I/O — como chamadas a APIs externas, bancos com drivers reativos (ex.: MongoDB Reactive, R2DBC), ou comunicação via filas de mensagens. Porém, aplicações baseadas em bancos tradicionais via JDBC não se beneficiam do modelo reativo, pois JDBC é bloqueante.

De forma geral, aplicações reativas tendem a ser mais escaláveis, leves, rápidas e tolerantes à sobrecarga, tornando-se especialmente adequadas para arquiteturas distribuídas e ambientes de alta concorrência.

Fluxo Reactive Spring:

A thread entra no pipeline reativo, executa um pedaço do trabalho e entrega a operação de I/O para o sistema. Depois disso, ela fica livre para atender
outras requisições.

Quando o I/O termina, o pipeline reativo emite um evento indicando que os dados retornaram. Uma thread disponível entra novamente no pipeline no ponto correto, pega esses dados e continua o fluxo até o fim.

Este trabalho consiste num conjunto de 4 API's, onde hello_api retorna "hello", world_api retorna "world", hello_world_api retorna o resultado de hello_api e world_api "hello world!". Os dados são persistidos através de um serviço de banco de dados db_service.

#### Explicação do código:

##### Microserviço db_service

A classe CallLog representa a entidade armazenada na tabela call_log do banco. Ela contém campos mapeados para as colunas da tabela por meio das anotações @Table e @Column. Os atributos são id que é a chave primária gerenciada pelo banco, serviceName que registra qual microserviço realizou a chamada, response que armazena a resposta recebida pelo serviço, timestamp que registra automaticamente o momento da criação do registro. O construtor define serviceName e response recebidos como parâmetros e define o timestamp como LocalDateTime.now() no momento da criação do log. A classe também possui getters e setters para permitir leitura e alteração dos campos pelo Spring e pelo mecanismo reativo.


A classe CallLogController expõe endpoints HTTP para interação com o serviço de logs. Ela é anotada com @RestController e atende na rota /logs. O método saveLog usa o verbo POST, recebe service e response como parâmetros de requisição e delega ao serviço CallLogService a tarefa de salvar o registro, retornando um Mono que representa a operação assíncrona. O método listLogs utiliza o verbo GET e retorna um Flux contendo todos os logs armazenados, também de forma reativa. O controller atua apenas como camada de exposição HTTP, mantendo a lógica de negócios no serviço.


A interface DbServiceRepository estende ReactiveCrudRepository, fornecendo operações reativas padrão como salvar, buscar e listar entidades. Ela trabalha com objetos do tipo CallLog e usa Long como tipo da chave primária. O método findAll retorna um Flux contendo todos os logs registrados, permitindo consumo assíncrono e não bloqueante. O repositório é gerenciado automaticamente pelo Spring Data R2DBC.


A classe CallLogService contém a lógica de negócios relacionada ao registro e consulta dos logs. Ela recebe o repositório via injeção de dependência no construtor. O método saveLog cria um novo CallLog com os parâmetros fornecidos e delega a persistência ao repositório, retornando um Mono representando a operação assíncrona. O método listLogs apenas devolve o resultado de repository.findAll, permitindo que o controller liste todos os registros sem expor diretamente a camada de dados.


A classe DbServiceApplication contém o método main responsável por iniciar o microserviço. Ela está anotada com @SpringBootApplication, que ativa automaticamente o escaneamento de componentes, a configuração automática e a inicialização do contexto Spring. Ao executar SpringApplication.run, o serviço db_service é iniciado e exposto para receber requisições.


O arquivo application.properties contém as configurações base do microserviço. O nome da aplicação é definido como db_service, a porta de execução como 8083, e as credenciais do banco são lidas via variáveis de ambiente. A URL utiliza o protocolo r2dbc para permitir comunicação reativa com PostgreSQL. A propriedade spring.sql.init.mode=always garante que o esquema será inicializado sempre que o serviço subir.


O arquivo application-docker.properties define as mesmas configurações básicas porém com credenciais fixas, adequadas para execução dentro de containers Docker. O perfil docker utiliza usuário postgres e senha postgres. Também define spring.sql.init.platform e a localização do schema.sql, garantindo criação da tabela ao rodar no ambiente de container.


O arquivo schema.sql contém a instrução de criação da tabela call_log. Ele define a coluna id como SERIAL e chave primária, service_name como texto obrigatório de até 50 caracteres, response como texto obrigatório de até 255 caracteres, e timestamp como campo obrigatório para registrar o momento da criação do log. O comando inclui IF NOT EXISTS para evitar erros caso a tabela já exista.

##### Microserviço hello_api

A classe WebClientConfig é responsável por configurar o cliente HTTP reativo que permitirá a comunicação com o microserviço db_service. ##### docker-compose.yml

O arquivo docker-compose.yml é o responsável por orquestrar todos os serviços do ecossistema de microserviços, definindo como cada container deve ser construído, configurado, iniciado e conectado dentro do ambiente Docker. Ele cria uma infraestrutura completa, incluindo banco de dados, serviços independentes e o serviço agregador final.

O serviço postgres é o primeiro a ser definido. Ele utiliza a imagem oficial do PostgreSQL e configura um banco de dados chamado db_service_pg, com usuário e senha definidos. Esse container expõe a porta 5432 para o host e inclui um healthcheck que utiliza o comando pg_isready para garantir que o banco só será considerado disponível após estar realmente pronto para conexões. Ele também é colocado na rede interna msnet, que é onde todos os outros serviços irão se comunicar.

Em seguida, o serviço db-service é construído a partir do diretório local ./db_service, onde está o Dockerfile correspondente. Esse serviço depende explicitamente do PostgreSQL, graças ao depends_on com a condição service_healthy, garantindo que ele só inicie quando o banco estiver funcionando. Ele expõe a porta 8083 e recebe variáveis de ambiente que ativam o perfil docker do Spring Boot e fornecem as credenciais do banco.

O serviço hello-api representa o microserviço responsável por retornar a string “Hello”. Ele também é construído a partir de seu diretório próprio e expõe a porta 8081. Recebe a variável DB_SERVICE_URL, apontando para o endpoint interno do serviço de banco de dados. Além disso, possui um healthcheck baseado no comando wget, que verifica periodicamente se o endpoint /hello está respondendo corretamente. Isso permite que o Compose saiba quando o serviço está realmente funcional.

De forma semelhante, o serviço world-api é o microserviço que retorna “World”. Ele é construído localmente, expõe a porta 8082 e utiliza a mesma variável DB_SERVICE_URL. Assim como o hello-api, possui um healthcheck que valida o endpoint /world, garantindo que esteja pronto antes de ser utilizado pelos outros serviços.

O serviço hello-world-api é o serviço agregador da aplicação, responsável por chamar o hello-api e o world-api, combinar as respostas e retornar “Hello World”. Ele é construído a partir do diretório ./hello_world_api e depende das duas APIs anteriores, mas com uma diferença importante: sua dependência utiliza condition: service_healthy, o que significa que ele só será iniciado quando ambos os serviços estiverem saudáveis e respondendo no respectivo healthcheck. Esse serviço recebe as variáveis HELLO e WORLD, contendo as URLs internas para acessar cada microserviço dentro da rede Docker. Ele expõe a porta 8080, que é a porta final por onde a aplicação agregada pode ser acessada externamente.

Por fim, o arquivo define uma rede chamada msnet, com driver do tipo bridge. Essa rede funciona como um ambiente isolado onde todos os containers podem se comunicar utilizando seus nomes de serviço como hostname, possibilitando que chamadas como http://hello-api:8081 funcionem normalmente, independentemente do host físico.

De maneira geral, esse docker-compose cria todo o ambiente necessário para que os microserviços funcionem de forma integrada, automatiza a ordem de inicialização, garante que serviços só iniciem quando suas dependências estiverem saudáveis, padroniza as variáveis de ambiente e cria uma máquina completa, isolada e reproduzível para rodar todo o sistema.
